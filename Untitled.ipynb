```python
from IPython.display import HTML

HTML("""
<div style="width:420px;margin:auto;padding:20px;border-radius:15px;
background:#111;color:white;font-family:Arial;text-align:center;
box-shadow:0px 0px 15px gray;">

<h2 style="color:#00ff66;"> Rock Paper Scissors ✂️</h2>
<p style="color:#ccc;">Click a button to play against computer</p>

<div style="margin:15px 0;">
<button style="padding:12px 18px;margin:5px;border-radius:10px;
border:none;background:#ff4444;color:white;font-size:16px;
cursor:pointer;"
onclick="play('Rock')">🪨 Rock</button>

<button style="padding:12px 18px;margin:5px;border-radius:10px;
border:none;background:#4488ff;color:white;font-size:16px;
cursor:pointer;"
onclick="play('Paper')">📄 Paper</button>

<button style="padding:12px 18px;margin:5px;border-radius:10px;
border:none;background:#ffaa00;color:white;font-size:16px;
cursor:pointer;"
onclick="play('Scissors')">✂️ Scissors</button>
</div>

<div style="background:#222;padding:15px;border-radius:10px;">
<h3 style="margin:0;color:#fff;">📌 Result</h3>
<p id="result" style="font-size:16px;color:#0f0;margin-top:10px;">
Make your move!
</p>
</div>

<div style="margin-top:15px;background:#222;padding:12px;border-radius:10px;">
<h3 style="margin:0;color:#fff;">🏆 Score Board</h3>
<p style="margin:8px 0;font-size:16px;">
You: <b id="userScore">0</b><br>
Computer: <b id="compScore">0</b>
</p>
</div>

<button style="margin-top:18px;padding:10px 20px;border-radius:10px;
border:none;background:#00ff66;color:black;font-size:16px;
cursor:pointer;"
onclick="restart()">🔄 Restart</button>

<script>
let userScore = 0;
let compScore = 0;

function play(user){
    const choices = ["Rock","Paper","Scissors"];
    const computer = choices[Math.floor(Math.random()*3)];
    let result = "";

    if(user === computer){
        result = "🤝 Tie!";
    }
    else if(
        (user === "Rock" && computer === "Scissors") ||
        (user === "Paper" && computer === "Rock") ||
        (user === "Scissors" && computer === "Paper")
    ){
        result = "✅ You Win!";
        userScore++;
    }
    else{
        result = "❌ Computer Wins!";
        compScore++;
    }

    document.getElementById("result").innerHTML =
        "You chose: <b>" + user + "</b><br>" +
        "Computer chose: <b>" + computer + "</b><br><br>" +
        "<b style='color:yellow;'>" + result + "</b>";

    document.getElementById("userScore").innerHTML = userScore;
    document.getElementById("compScore").innerHTML = compScore;
}

function restart(){
    userScore = 0;
    compScore = 0;
    document.getElementById("userScore").innerHTML = 0;
    document.getElementById("compScore").innerHTML = 0;
    document.getElementById("result").innerHTML =
        "Game Restarted! 🎮 Make your move!";
}
</script>
</div>
""")
```
